# Sparsity of matrices for noise simulations

## Sparsity of noisy operation matrices

````python
from quantumsim_performante.quantumsim_performante import *

N = 10

print(f"Elements in operation matrix: {2**N*2**N}")

circuit = NoisyCircuit(N)
circuit.noisy_cnot(0, 1)
circuit.cnot(0, 1)
print("CNOT")
print(f"  > Noisy  NNZ: {circuit.operations[0].nnz}")
print(f"  > Normal NNZ: {circuit.operations[1].nnz}")

circuit = NoisyCircuit(N)
circuit.noisy_hadamard(0)
circuit.hadamard(0)
print("HADAMARD")
print(f"  > Noisy  NNZ: {circuit.operations[0].nnz}")
print(f"  > Normal NNZ: {circuit.operations[1].nnz}")

circuit = NoisyCircuit(N)
circuit.noisy_pauli_x(0)
circuit.pauli_x(0)
print("PAULI X")
print(f"  > Noisy  NNZ: {circuit.operations[0].nnz}")
print(f"  > Normal NNZ: {circuit.operations[1].nnz}")

circuit = NoisyCircuit(N)
circuit.noisy_phase(10, 0)
circuit.phase(10, 0)
print("PHASE")
print(f"  > Noisy  NNZ: {circuit.operations[0].nnz}")
print(f"  > Normal NNZ: {circuit.operations[1].nnz}")
````

Remember, every additional qubits makes the matrices twice as sparse. This can be found in `QuantumSimComputationalImprovements.ipynb`

## Sparsity of noise matrices

Talking about these three noise matrices:

```python
self.noisy_operations_state_prep: sparse.coo_matrix = []
self.noisy_operations_incoherent: sparse.coo_matrix = []
self.noisy_operations_readout: sparse.coo_matrix = []
```

These noise matrices are multiplied with the `StateVector` during execution or measurement

````python
print("Sparsity noise matrices: ")

for N in range(3, 7):
    elements = 2**N*2**N
    print(f"    > Elements in dense noise matrix: {elements}")
    circuit = NoisyCircuit(N)
    circuit.add_noisy_operation_state_prep(0.07, 1)
    circuit.add_noisy_operation_incoherent(0.003, 0.004, 0.005, 1)
    circuit.add_noisy_operation_readout(0.04, 0.03, 0)

    print(f"\t> noisy_operations_state_prep: {circuit.noisy_operations_state_prep[0].nnz/elements * 100}%")
    print(f"\t> noisy_operations_incoherent: {circuit.noisy_operations_incoherent[0].nnz/elements * 100}%")
    print(f"\t> noisy_operations_readout: {circuit.noisy_operations_readout[0].nnz/elements * 100}%\n")
````

Just like with the unitary operational matrices, every additional qubits makes the noise matrices twice as sparse.
